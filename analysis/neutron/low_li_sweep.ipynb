```python
import openmc
import numpy as np
import matplotlib.pyplot as plt
```

```python
folders = ["li_2_0", "li_3_0", "li_4_0", "li_5_0", "li_6_0", "li_7_0", "li_7_6"]

means = []
std_devs = []

for folder in folders:
    sp = openmc.StatePoint(f"{folder}/statepoint.100.h5")
    mean = sp.get_tally(name="TBR").mean.flatten()[0]
    std_dev = sp.get_tally(name="TBR").std_dev.flatten()[0]
    means.append(mean)
    std_devs.append(std_dev)


means = np.array(means)
std_devs = np.array(std_devs)
```

```python
concentractions = [2, 3, 4, 5, 6, 7, 7.6]
xlabels = [f"{c}%" for c in concentractions]
plt.errorbar(concentractions, means, yerr=std_devs, fmt='o', capsize=5)
plt.xticks(concentractions, xlabels)
plt.xlabel("Lithium Concentration (wt%)")
plt.ylabel("TBR")
plt.grid()
```
